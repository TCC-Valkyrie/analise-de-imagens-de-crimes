```python
from ultralytics import YOLO

def retrain_yolo_model():
    model = YOLO("best.pt")
    
    results = model.train(
        data="dataset_faces/data.yaml",
        epochs=10,
        imgsz=640,
        batch=16,
        device=0
    )

    print("Retraining complete!")

if __name__ == "__main__":
    retrain_yolo_model()
```
