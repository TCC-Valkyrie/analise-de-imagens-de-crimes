```python
import torch
import torch.nn as nn
from ultralytics import YOLO
```

```python
# Testing CUDA availability
print("CUDA Available: ", torch.cuda.is_available())
print("GPU Name: ", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "None")
```

```python
# 0. Set random seed for reproducibility
torch.manual_seed(42)

# 1. Load a pre-trained YOLOv8 nano model
model = YOLO("yolov8m.pt") 

if torch.cuda.is_available():
    model.to("cuda")
```

```python
def alterar_ativacao(modulo, ativacao_antiga, nova_ativacao):
    for nome, filho in modulo.named_children():
        if isinstance(filho, ativacao_antiga):
            setattr(modulo, nome, nova_ativacao())
        else:
            alterar_ativacao(filho, ativacao_antiga, nova_ativacao)
```

```python
alterar_ativacao(model.model, nn.SiLU, nn.Mish)
```

```python
results = model.train(
    data="dataset/data.yaml",
    epochs=25,
    imgsz=640,
    batch=16,
    lr0=0.001,
    optimizer="RAdam",


    project="crime-model-epochs-25-teste",
    name="training_results"
)
```

```python
import yaml
from ultralytics import YOLO

def main():
    yaml_path = "runs/detect/crime-model-epochs-15-ReLU,RAdam/training_RAdam_15_epochs-ReLU/args.yaml"
    
    with open(yaml_path, "r", encoding="utf-8") as f:
        config = yaml.safe_load(f)
    
    model = YOLO(config.get("model", "yolov8m.pt"))
    
    results = model.train(**config)
    
    print("Training finished successfully!")

if __name__ == "__main__":
    main()
```
