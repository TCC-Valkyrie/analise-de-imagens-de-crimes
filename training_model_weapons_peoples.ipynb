```python
# Deep Learning Libraries
import torch
from ultralytics import YOLO
import torch.nn as nn
```

```python
# Testing CUDA availability
print("CUDA Available: ", torch.cuda.is_available())
print("GPU Name: ", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "None")
```

```python
# 0. Set random seed for reproducibility
torch.manual_seed(42)

# 1. Load a pre-trained YOLOv8 medium model
model = YOLO("yolov8m.pt") 


if torch.cuda.is_available():
    model.to("cuda")  # Move the model to GPU if available
```

```python
def alterar_ativacao(modulo, ativacao_antiga, nova_ativacao):
    for nome, filho in modulo.named_children():
        if isinstance(filho, ativacao_antiga):
            # Substitui a camada antiga pela nova
            setattr(modulo, nome, nova_ativacao())
        else:
            # Busca recursivamente nas sub-camadas
            alterar_ativacao(filho, ativacao_antiga, nova_ativacao)
```

```python
alterar_ativacao(model.model, nn.SiLU, nn.GELU)
```

```python
alterar_ativacao(model.model, nn.SiLU, nn.ReLU)
```

```python
# Troca todas as SiLU por Mish
alterar_ativacao(model.model, nn.SiLU, nn.Mish)
```

```python
results = model.train(
    data="dataset/data.yaml",
    epochs=15,
    imgsz=640,
    batch=16,
    optimizer="RAdam",
    lr0=0.001,
    project="crime-model-epochs-15-Mish-RAdam",              
    name="training_RAdam_15_epochs-Mish",
)
```
